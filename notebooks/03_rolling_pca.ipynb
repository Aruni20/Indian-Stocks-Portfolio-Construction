# 03. Rolling PCA and rebalancing

This notebook studies how a one-year trailing estimation window behaves under four rebalance schedules.

## What rolling means

At each rebalance date, only the previous year is used to estimate the covariance matrix and PCA loadings. The resulting weights are held until the next rebalance.

## Visual interpretation

The committed rolling-return figure compares monthly, quarterly, half-yearly, and yearly schedules. More frequent rebalancing reacts faster but can create more turnover; slower schedules are more stable but allow weights to become stale.

## Look-ahead check

The one-year window ends before each rebalance. No future observation is used to construct the next portfolio.

```python
# Visual comparison: rolling PCA schedules
import pandas as pd
import matplotlib.pyplot as plt
rolling = pd.read_csv(ROOT / "outputs" / "rolling_pca_yearly_returns_sector_average.csv")
rolling.plot(x=rolling.columns[0], marker="o", title="Rolling PCA annual return by rebalance frequency")
plt.ylabel("Annual return")
plt.tight_layout()
plt.show()
```
