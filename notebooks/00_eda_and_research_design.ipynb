## Research questions

1. How do the selected stocks and sectors differ in return and risk?
2. Does covariance-aware allocation improve on equal weighting?
3. Does a PCA eigenportfolio produce a useful alternative to MPT portfolios?
4. How sensitive are results to monthly, quarterly, half-yearly, and yearly rolling rebalancing?

The analysis is long-only: weights are non-negative and normalized to sum to one.

## Interpretation

Equal weight is the transparent benchmark. Minimum-risk uses the covariance structure to reduce portfolio variance. Optimum-risk maximizes estimated excess return per unit of risk. The eigenportfolio uses PCA loadings as a factor-based candidate and normalizes absolute loadings for a long-only implementation.

The fixed experiment follows the paper's train/test logic. The rolling experiment prevents look-ahead by using only the preceding one year to form the next rebalance weights. The results are descriptive historical estimates, not guarantees.
